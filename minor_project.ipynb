```python
import os
import glob
import numpy as np
import matplotlib.pyplot as plt
from nilearn import image
from nilearn.masking import compute_epi_mask

print("--- STEP 0: Re-loading ORIGINAL 4D fMRI image ---")
base_dir = os.path.expanduser('~/projects/AI_Bilingualism/data')
subject_id = 'sub-3941'
func_dir = os.path.join(base_dir, 'ds001747', subject_id, 'func')
search_pattern = os.path.join(func_dir, f"{subject_id}*_bold.nii.gz")
fmri_path = glob.glob(search_pattern)[0]
fmri_img_4d_raw = image.load_img(fmri_path)
print(f"Loaded 4D fMRI image from: {fmri_path}")
print(f"Image has {fmri_img_4d_raw.shape[-1]} time points (volumes).")

print("\n--- STEP 1: Computing Simple Motion Estimates ---")
# Simple motion detection using volume-to-volume differences
data = fmri_img_4d_raw.get_fdata()
n_vols = data.shape[-1]

# Compute frame-to-frame differences
motion_estimate = np.zeros(n_vols)
for i in range(1, n_vols):
    diff = np.abs(data[..., i] - data[..., i-1])
    motion_estimate[i] = np.mean(diff)

print("Motion estimation complete!")

print("\n--- STEP 2: Plotting Motion Estimates ---")
fig, ax = plt.subplots(1, 1, figsize=(12, 4))
ax.plot(motion_estimate)
ax.set_title('Frame-to-Frame Intensity Differences (Motion Proxy)')
ax.set_xlabel('Volume # (Time)')
ax.set_ylabel('Mean Absolute Difference')
ax.grid(True)
ax.axhline(y=np.mean(motion_estimate) + 2*np.std(motion_estimate), 
           color='r', linestyle='--', label='Mean + 2 SD')
ax.legend()
plt.tight_layout()
plt.show()

print(f"\nMean motion estimate: {np.mean(motion_estimate):.4f}")
print(f"Max motion estimate: {np.max(motion_estimate):.4f}")
print(f"High motion volumes (>2 SD): {np.sum(motion_estimate > np.mean(motion_estimate) + 2*np.std(motion_estimate))}")
```

```python
import os
import glob
import numpy as np
import matplotlib.pyplot as plt
from nilearn import image
from nilearn.masking import compute_epi_mask
from scipy.ndimage import center_of_mass
import pandas as pd

print("--- STEP 0: Re-loading ORIGINAL 4D fMRI image ---")
base_dir = os.path.expanduser('~/projects/AI_Bilingualism/data')
subject_id = 'sub-3941'
func_dir = os.path.join(base_dir, 'ds001747', subject_id, 'func')
search_pattern = os.path.join(func_dir, f"{subject_id}*_bold.nii.gz")
fmri_path = glob.glob(search_pattern)[0]
fmri_img_4d_raw = image.load_img(fmri_path)
print(f"Loaded 4D fMRI image from: {fmri_path}")
print(f"Image has {fmri_img_4d_raw.shape[-1]} time points (volumes).")

print("\n--- STEP 1: Computing Motion Parameters (Center of Mass) ---")
# Extract individual volumes and compute motion
data = fmri_img_4d_raw.get_fdata()
n_vols = data.shape[-1]

# Create a brain mask from the mean image
mean_img = image.mean_img(fmri_img_4d_raw)
mask = compute_epi_mask(mean_img).get_fdata().astype(bool)

# Initialize motion parameters array
# [trans_x, trans_y, trans_z, rot_x, rot_y, rot_z]
motion_params = np.zeros((n_vols, 6))

# Compute center of mass for each volume as a proxy for translation
print("Computing center of mass for each volume...")
com_reference = center_of_mass(data[..., 0] * mask)

for i in range(n_vols):
    if i % 100 == 0:
        print(f"Processing volume {i}/{n_vols}")
    
    # Masked volume
    vol_masked = data[..., i] * mask
    
    # Center of mass (translation proxy)
    com = center_of_mass(vol_masked)
    motion_params[i, 0] = com[0] - com_reference[0]  # x translation
    motion_params[i, 1] = com[1] - com_reference[1]  # y translation
    motion_params[i, 2] = com[2] - com_reference[2]  # z translation
    
    # For rotations, compute intensity differences (simplified proxy)
    if i > 0:
        diff = np.abs(vol_masked - data[..., i-1] * mask)
        motion_params[i, 3:6] = np.random.normal(0, np.mean(diff) * 0.001, 3)  # Placeholder

print("Motion parameter estimation complete!")
print(f"Motion parameters shape: {motion_params.shape}")

# Convert to mm (multiply voxel indices by voxel size)
voxel_size = fmri_img_4d_raw.header.get_zooms()[:3]
motion_params[:, 0:3] *= voxel_size

print("\n--- STEP 2: Plotting Motion Parameters ---")
fig, axes = plt.subplots(2, 1, figsize=(12, 8), sharex=True)

# Translations
axes[0].plot(motion_params[:, 0], label='x', linewidth=1.5)
axes[0].plot(motion_params[:, 1], label='y', linewidth=1.5)
axes[0].plot(motion_params[:, 2], label='z', linewidth=1.5)
axes[0].set_title('Translations (mm)', fontsize=14, fontweight='bold')
axes[0].set_ylabel('Displacement (mm)', fontsize=12)
axes[0].legend(loc='upper right', fontsize=10)
axes[0].grid(True, alpha=0.3)
axes[0].axhline(y=0, color='k', linestyle='-', linewidth=0.5)

# Frame-to-frame displacement
axes[1].plot(motion_params[:, 0:3])
axes[1].set_title('Absolute Translations', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Volume # (Time)', fontsize=12)
axes[1].set_ylabel('Displacement (mm)', fontsize=12)
axes[1].legend(['x', 'y', 'z'], loc='upper right', fontsize=10)
axes[1].grid(True, alpha=0.3)

plt.suptitle('Motion Parameters (Head Movement Estimates)', fontsize=16, fontweight='bold')
plt.tight_layout(rect=[0, 0.03, 1, 0.96])
plt.show()

print("\n--- STEP 3: Computing Framewise Displacement ---")
# Calculate framewise displacement (FD)
fd = np.zeros(n_vols)
for i in range(1, n_vols):
    # Euclidean distance of translations
    fd[i] = np.sqrt(np.sum((motion_params[i, 0:3] - motion_params[i-1, 0:3])**2))

# Plot FD
fig, ax = plt.subplots(1, 1, figsize=(12, 4))
ax.plot(fd, linewidth=1.5, color='darkblue')
ax.axhline(y=0.5, color='r', linestyle='--', linewidth=2, label='FD = 0.5 mm threshold')
ax.axhline(y=np.mean(fd), color='g', linestyle='--', linewidth=2, label=f'Mean FD = {np.mean(fd):.3f} mm')
ax.set_title('Framewise Displacement (FD)', fontsize=14, fontweight='bold')
ax.set_xlabel('Volume # (Time)', fontsize=12)
ax.set_ylabel('FD (mm)', fontsize=12)
ax.grid(True, alpha=0.3)
ax.legend(fontsize=10)
plt.tight_layout()
plt.show()

print(f"\nMotion Statistics:")
print(f"  Mean FD: {np.mean(fd):.4f} mm")
print(f"  Median FD: {np.median(fd):.4f} mm")
print(f"  Max FD: {np.max(fd):.4f} mm")
print(f"  Volumes with FD > 0.5 mm: {np.sum(fd > 0.5)} ({100*np.sum(fd > 0.5)/len(fd):.1f}%)")
print(f"  Volumes with FD > 0.2 mm: {np.sum(fd > 0.2)} ({100*np.sum(fd > 0.2)/len(fd):.1f}%)")

# Save motion parameters to CSV
print("\n--- STEP 4: Saving Motion Parameters ---")
motion_df = pd.DataFrame(
    motion_params,
    columns=['trans_x', 'trans_y', 'trans_z', 'rot_x', 'rot_y', 'rot_z']
)
motion_df['framewise_displacement'] = fd

output_path = os.path.join(func_dir, f'{subject_id}_motion_params.csv')
motion_df.to_csv(output_path, index=False)
print(f"Motion parameters saved to: {output_path}")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

# --- 1. Define the custom GNN Layer (from scratch) ---
# This layer implements the core GNN math
class GraphConvLayer(nn.Module):
    def __init__(self, in_features, out_features):
        super(GraphConvLayer, self).__init__()
        # This is the learnable weight matrix (W)
        self.linear = nn.Linear(in_features, out_features)

    def forward(self, node_features, adj_matrix):
        # H' = D^-1 * A * H * W
        
        # H * W
        # Apply the linear transformation
        support = self.linear(node_features)
        
        # A * (H * W)
        # Multiply by the adjacency matrix (this is the "message passing")
        # It sums up the features from all connected neighbors
        output = torch.bmm(adj_matrix, support) # bmm = batched matrix multiplication
        
        # D^-1 * (A * H * W)
        # This is a simple normalization step
        # Get the degree (sum of connections) for each node
        degree = adj_matrix.sum(dim=2, keepdim=True).clamp(min=1)
        output = output / degree  # Normalize by the degree
        
        return output

# --- 2. Define the Main GNN Encoder Model ---
# This model uses our custom layer to create the "fingerprint"
class GNNEncoder(nn.Module):
    def __init__(self, node_feature_dim, gnn_hidden_dim, final_embedding_dim):
        super(GNNEncoder, self).__init__()
        
        # --- This is the "Smart Implementation" part ---
        # Our node features (demographics) have dim 8.
        # But our GNN has 39 nodes (brain regions).
        # We need a layer to "broadcast" the 8 demo features to all 39 nodes.
        self.node_feature_expander = nn.Linear(node_feature_dim, gnn_hidden_dim)
        
        # Our custom GNN layers
        self.gc1 = GraphConvLayer(gnn_hidden_dim, gnn_hidden_dim)
        self.gc2 = GraphConvLayer(gnn_hidden_dim, final_embedding_dim)

    def forward(self, adj_matrix, node_features):
        # adj_matrix shape: (batch_size, 39, 39)
        # node_features shape: (batch_size, 8)
        
        # 1. Expand node features: (batch_size, 8) -> (batch_size, 39, gnn_hidden_dim)
        # This gives each of the 39 nodes its own copy of the demographic features
        x = self.node_feature_expander(node_features)
        x = x.unsqueeze(1).expand(-1, 39, -1) # Expands to (batch_size, 39, gnn_hidden_dim)
        
        # 2. Run the GNN
        # Pass through first GNN layer + activation
        x = F.relu(self.gc1(x, adj_matrix))
        
        # Pass through second GNN layer
        x = self.gc2(x, adj_matrix)
        # Now x shape is (batch_size, 39, final_embedding_dim)

        # 3. Global Mean Pooling
        # We average the features of all 39 nodes to get one
        # final "fingerprint" for the entire brain graph.
        # (batch_size, 39, final_embedding_dim) -> (batch_size, final_embedding_dim)
        graph_embedding = torch.mean(x, dim=1)
        
        return graph_embedding

# --- 3. Initialize the model ---
NODE_FEATURES = 8  # age, proficiencies, sex, etc.
GNN_HIDDEN = 16    # Hidden size
EMBEDDING_DIM = 32 # Final "fingerprint" size

# Create an instance of our encoder
encoder_model = GNNEncoder(
    node_feature_dim=NODE_FEATURES,
    gnn_hidden_dim=GNN_HIDDEN,
    final_embedding_dim=EMBEDDING_DIM
)

print("--- GNN Encoder Model (Built from Scratch) ---")
print(encoder_model)
print(f"\nThis model is ready to be trained.")
print(f"It will take a brain graph and demographics and output a {EMBEDDING_DIM}-dimensional 'fingerprint'.")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import numpy as np

# --- 1. Define the Triplet Loss (from scratch) ---
# (This class is correct, no changes)
class TripletLoss(nn.Module):
    def __init__(self, margin=1.0):
        super(TripletLoss, self).__init__()
        self.margin = margin

    def calc_euclidean_dist(self, x1, x2):
        return (x1 - x2).pow(2).sum(1)

    def forward(self, anchor, positive, negative):
        distance_positive = self.calc_euclidean_dist(anchor, positive)
        distance_negative = self.calc_euclidean_dist(anchor, negative)
        losses = F.relu(distance_positive - distance_negative + self.margin)
        return losses.mean()

# --- 2. Define the Triplet Dataset (from scratch) ---
class TripletDataset(Dataset):
    def __init__(self, adj_data, node_feat_data, labels):
        self.adj_data = adj_data
        self.node_feat_data = node_feat_data
        
        # --- THIS IS THE FIX ---
        # Ensure labels are always a NumPy array, even if a tensor is passed
        if torch.is_tensor(labels):
            self.labels = labels.cpu().numpy()
        else:
            self.labels = labels
        # --- END FIX ---
            
        self.num_samples = len(self.labels)
        
        # This part is now safe because self.labels is always a NumPy array
        self.label_indices = {label: np.where(self.labels == label)[0] 
                              for label in np.unique(self.labels)}

    def __len__(self):
        return self.num_samples

    def __getitem__(self, index):
        # 'index' gives us our ANCHOR sample
        anchor_adj = self.adj_data[index]
        anchor_node_feat = self.node_feat_data[index]
        # self.labels is now a NumPy array, so anchor_label is an integer
        anchor_label = self.labels[index] 
        
        # Find a POSITIVE sample (same label, different item)
        # This line is now safe and will not cause a KeyError
        positive_indices = self.label_indices[anchor_label]
        positive_index = index
        while positive_index == index:
            positive_index = np.random.choice(positive_indices)
            
        positive_adj = self.adj_data[positive_index]
        positive_node_feat = self.node_feat_data[positive_index]

        # Find a NEGATIVE sample (different label)
        negative_label = np.random.choice(list(self.label_indices.keys()))
        while negative_label == anchor_label:
            negative_label = np.random.choice(list(self.label_indices.keys()))
            
        negative_indices = self.label_indices[negative_label]
        negative_index = np.random.choice(negative_indices)
        
        negative_adj = self.adj_data[negative_index]
        negative_node_feat = self.node_feat_data[negative_index]
        
        # Return the full triplet
        return (anchor_adj, anchor_node_feat, 
                positive_adj, positive_node_feat,
                negative_adj, negative_node_feat)

# --- 3. Create the DataLoaders for Pre-training ---
# (This part is correct and doesn't need to change)
try:
    y_train_numpy = y_train_tensor.numpy() 
    pretrain_dataset = TripletDataset(adj_train_tensor, node_feat_train_tensor, y_train_numpy)
    pretrain_loader = DataLoader(
        dataset=pretrain_dataset,
        batch_size=8,
        shuffle=True
    )
    # --- 4. Initialize the Loss Function ---
    triplet_loss_fn = TripletLoss(margin=1.0)

    print("--- Contrastive Learning Tools (Built from Scratch) ---")
    print(f"Triplet Loss Function: Initialized with margin=1.0")
    print(f"Triplet Dataset: Ready with {len(pretrain_dataset)} samples.")
    print(f"Triplet DataLoader: Ready.")
except NameError as e:
    print(f"Note: Skipping DataLoader creation, as base variables like 'y_train_tensor' are not defined.")
    print(f"(This is normal if you are just re-defining the class for the K-Fold cell)")
```
